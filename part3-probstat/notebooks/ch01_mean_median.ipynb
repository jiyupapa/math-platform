## 0. 준비

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

rng = np.random.default_rng(42)
print('준비 완료!')
```

## 1. 평균 — 직접 계산 vs numpy

평균 = 다 더해서 개수로 나누기. 웹 챕터의 예제 [4, 7, 7, 9, 13] 으로 확인합니다.

```python
data = [4, 7, 7, 9, 13]

mean_manual = sum(data) / len(data)   # 직접: 합 ÷ 개수
mean_numpy = np.mean(data)            # numpy 한 줄

print('직접 계산 :', mean_manual)
print('np.mean  :', mean_numpy)
```

## 2. 중앙값 — 정렬 먼저!

중앙값은 **크기순으로 정렬한 뒤** 한가운데 값입니다. 개수가 짝수면 가운데 두 값의 평균.

```python
odd_data = [7, 1, 5]            # 홀수 개 — 정렬하면 [1, 5, 7], 가운데는 5
even_data = [2, 4, 6, 10]       # 짝수 개 — 가운데 두 값 4, 6 의 평균 = 5

print('정렬       :', sorted(odd_data))
print('중앙값(홀수):', np.median(odd_data))
print('중앙값(짝수):', np.median(even_data))
```

## 3. 최빈값 — 가장 자주 나오는 값

파이썬 표준 라이브러리 `collections.Counter` 로 각 값의 등장 횟수를 세면 됩니다.

```python
from collections import Counter

shoe_sizes = [250, 255, 260, 260, 265, 265, 265, 270, 270, 275]
counts = Counter(shoe_sizes)

print('값별 등장 횟수 :', dict(counts))
print('최빈값        :', counts.most_common(1)[0][0], 'mm')  # 가장 많이 팔린 사이즈
```

## 4. 이상치 실험 — "평균 연봉"의 함정

직원 4명(3000~4500만원)에 회장님(30억)을 넣으면 평균과 중앙값이 어떻게 달라질까요?

```python
employees = [3000, 3500, 4000, 4500]          # 만원 단위
with_boss = employees + [300000]              # 회장님 합류 (30억)

print('회장님 없이 → 평균:', np.mean(employees), ' 중앙값:', np.median(employees))
print('회장님 포함 → 평균:', np.mean(with_boss), ' 중앙값:', np.median(with_boss))
print()
print('평균은 3750 → 63000(6.3억)으로 폭등, 중앙값은 3750 → 4000으로 거의 그대로!')
```

## 5. 히스토그램 — 데이터의 모양 보기

학생 200명의 키를 만들어 히스토그램으로 그리고, 평균(빨강)·중앙값(파랑) 위치를 표시합니다.

🧪 **직접 실험**: `bins=15` 를 5나 40으로 바꿔 보세요 — 웹의 '계급 폭 슬라이더'와 같은 실험입니다.

```python
heights = rng.normal(170, 7, size=200)   # 평균 170, 표준편차 7 인 키 200개 (6장에서 배울 정규분포)

plt.figure(figsize=(8, 4))
plt.hist(heights, bins=15, color='#818cf8', edgecolor='white')
plt.axvline(np.mean(heights), color='#ef4444', linestyle='--', linewidth=2,
            label=f'평균 {np.mean(heights):.1f}')
plt.axvline(np.median(heights), color='#3b82f6', linestyle=':', linewidth=2,
            label=f'중앙값 {np.median(heights):.1f}')
plt.xlabel('키 (cm)')
plt.ylabel('학생 수')
plt.title('학생 200명의 키 — 대칭 분포라 평균 ≈ 중앙값')
plt.legend()
plt.show()
```

이번엔 **오른쪽 꼬리가 긴** 데이터(연봉 느낌)로 그려 봅니다 — 평균이 중앙값보다
오른쪽으로 밀려나는 것을 확인하세요.

```python
salaries = rng.lognormal(mean=8.2, sigma=0.5, size=500)   # 오른쪽 꼬리가 긴 분포

plt.figure(figsize=(8, 4))
plt.hist(salaries, bins=40, color='#fbbf24', edgecolor='white')
plt.axvline(np.mean(salaries), color='#ef4444', linestyle='--', linewidth=2,
            label=f'평균 {np.mean(salaries):.0f}')
plt.axvline(np.median(salaries), color='#3b82f6', linestyle=':', linewidth=2,
            label=f'중앙값 {np.median(salaries):.0f}')
plt.xlabel('연봉 (만원)')
plt.ylabel('사람 수')
plt.title('오른쪽 꼬리가 긴 분포 — 평균 > 중앙값')
plt.legend()
plt.show()
```

## 정리

- **평균** = 합 ÷ 개수 (무게중심) · **중앙값** = 정렬 후 가운데 · **최빈값** = 가장 잦은 값.
- **이상치** 하나에 평균은 폭등하지만 중앙값은 버팁니다 — "평균 연봉"을 조심하세요.
- 히스토그램에서 **오른쪽 꼬리가 길면 평균 > 중앙값**.
