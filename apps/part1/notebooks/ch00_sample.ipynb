## 1. 벡터 만들기

NumPy 에서 벡터는 그냥 숫자 배열입니다. `(3, 2)` 벡터를 만들어 봅시다.

```python
import numpy as np

a = np.array([3, 2])
print('벡터 a =', a)
print('x 성분 =', a[0], ', y 성분 =', a[1])
```

## 2. 벡터의 길이 (크기)

$$\lVert \vec{a} \rVert = \sqrt{x^2 + y^2}$$

직접 계산해 보고, NumPy 함수와 비교합니다.

```python
length_manual = (a[0]**2 + a[1]**2) ** 0.5
length_numpy = np.linalg.norm(a)

print('직접 계산:', length_manual)
print('NumPy   :', length_numpy)
```

## 3. 화살표로 그리기

Matplotlib 의 `quiver` 로 원점에서 뻗는 화살표를 그립니다.
웹 캔버스에서 본 그 화살표와 같은 것입니다.

```python
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic' 로 바꾸세요
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호(-) 깨짐 방지

fig, ax = plt.subplots(figsize=(5, 5))
ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='#6366f1', width=0.012)
ax.text(a[0] + 0.1, a[1] + 0.1, f'a = ({a[0]}, {a[1]})', color='#6366f1')

ax.set_xlim(-5, 5); ax.set_ylim(-5, 5)
ax.axhline(0, color='#94a3b8', lw=1); ax.axvline(0, color='#94a3b8', lw=1)
ax.grid(True, color='#eef2f7'); ax.set_aspect('equal')
ax.set_title('벡터 a')
plt.show()
```
