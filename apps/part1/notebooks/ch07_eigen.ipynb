```python
import numpy as np

A = np.array([[2.0, 1.0],
              [1.0, 2.0]])

vals, vecs = np.linalg.eig(A)
print('고유값 λ =', np.round(vals, 3))            # [3. 1.]
print('고유벡터(열) =\n', np.round(vecs, 3))      # 각 '열'이 고유벡터
```

## 1. Av = λv 가 정말 성립하나?

첫 번째 고유값·고유벡터로 양변을 직접 계산해 비교합니다.

```python
lam = vals[0]
v = vecs[:, 0]      # 첫째 열 = 첫째 고유벡터

left = A @ v        # 변환한 결과
right = lam * v     # 그냥 λ배 한 것
print('A v   =', np.round(left, 3))
print('λ v   =', np.round(right, 3))
print('같은가?', np.allclose(left, right))
```

## 2. 대각·삼각 행렬은 대각 성분이 고유값

계산해보면 대각 성분이 그대로 고유값으로 나옵니다.

```python
D = np.array([[5.0, 0.0],
              [0.0, 9.0]])
T = np.array([[4.0, 2.0],
              [0.0, 7.0]])   # 위 삼각행렬
print('대각 [[5,0],[0,9]] 고유값 =', np.round(np.linalg.eigvals(D), 2))   # [5. 9.]
print('삼각 [[4,2],[0,7]] 고유값 =', np.round(np.linalg.eigvals(T), 2))   # [4. 7.]
```

## 3. 직접 해보기 ✏️

`[[2,0],[0,3]]` 의 고유값이 2와 3 인지 확인하세요.

```python
M = np.array([[2.0, 0.0],
              [0.0, 3.0]])
vals = np.sort(np.linalg.eigvals(M).real)
print('고유값 =', np.round(vals, 2))
assert np.allclose(vals, [2.0, 3.0]), '다시 확인해 보세요!'
print('✅ 정답! 고유값 2, 3')
```
