```python
import numpy as np

# 행렬식 = 넓이 배율
A = np.array([[2, 0],
              [0, 3]])
print('det(A) =', round(np.linalg.det(A), 2))   # 6.0  (가로 2배 x 세로 3배)

B = np.array([[1, 2],
              [3, 4]])
print('det(B) =', round(np.linalg.det(B), 2))   # -2.0

# 2x2 공식 ad - bc 로 직접 확인
a, b, c, d = 1, 2, 3, 4
print('직접 계산 ad - bc =', a*d - b*c)           # -2
```

## 1. det = 0 이면 되돌릴 수 없다

행렬식이 0이면 공간이 납작해져 역행렬이 없습니다. `inv` 를 시도하면 에러가 나요.

> 새 이름 하나만: 이렇게 **행렬식이 0이라 되돌릴 수 없는 행렬**을 **특이행렬**(singular matrix)이라
> 불러요. 아래 에러 메시지에 나오는 영어 "Singular matrix"가 바로 그 이름입니다.
> (ch11에서 배울 '특이값'과는 별개 용어예요. 이름만 알아두면 충분해요.)

```python
S = np.array([[2, 4],
              [1, 2]])   # det = 2*2 - 4*1 = 0
print('det(S) =', round(np.linalg.det(S), 2))   # 0.0

try:
    np.linalg.inv(S)
except np.linalg.LinAlgError as e:
    print('역행렬 없음(특이행렬):', e)
```

## 2. 역행렬과 A⁻¹A = I

역행렬을 구하고, 원래 행렬과 곱하면 항등행렬 I 가 나오는지 확인합니다.

```python
A = np.array([[2.0, 1.0],
              [1.0, 3.0]])
A_inv = np.linalg.inv(A)
print('A^-1 =\n', np.round(A_inv, 3))

prod = A_inv @ A
print('\nA^-1 @ A =\n', np.round(prod, 6))   # 항등행렬 I
print('\nI 와 같은가?', np.allclose(prod, np.eye(2)))
```

## 3. 역행렬로 연립방정식 풀기: x = A⁻¹b

$A\vec{x}=\vec{b}$ 를 $\vec{x} = A^{-1}\vec{b}$ 로 풀고, `np.linalg.solve` 결과와 비교합니다.

```python
A = np.array([[1.0, 1.0],
              [1.0, -1.0]])
b = np.array([3.0, 1.0])

x_inv = np.linalg.inv(A) @ b      # 역행렬로
x_solve = np.linalg.solve(A, b)   # 소거법으로 (실전 권장)
print('x = A^-1 b :', x_inv)       # [2. 1.]
print('x = solve  :', x_solve)     # [2. 1.]
print('같은가?', np.allclose(x_inv, x_solve))
```

## 4. 직접 해보기 ✏️

`[[3,2],[1,4]]` 의 행렬식이 10 인지 확인하세요.

```python
M = np.array([[3, 2],
              [1, 4]])
d = round(np.linalg.det(M))
print('det =', d)
assert d == 10, '다시 확인해 보세요! (ad - bc = 3*4 - 2*1)'
print('✅ 정답! det = 10')
```
