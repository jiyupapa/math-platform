## 1단계. 세상 모든 걸 "숫자 나열"로 바꿔보기

컴퓨터는 '사과'라는 글자를 이해하지 못해요.
대신 **특징을 숫자로 적어주면** 알아듣습니다.

예를 들어 과일을 `[ 빨간정도, 둥근정도, 단맛 ]` 으로 적어보는 거죠.
이렇게 숫자를 줄 세운 걸 **벡터(vector)** 라고 부릅니다.

어렵게 생각 마세요. 그냥 *"특징을 숫자로 줄세운 것"* 입니다.

```python
# ====================================================================
# [1번 셀] 일상 물건을 숫자로 적어보기
# ====================================================================
# [ 빨간정도, 둥근정도, 단맛 ] 을 0~5로 매겨봤어요.
# 정답은 없어요. 그냥 느낌대로 적은 겁니다.
사과     = np.array([5, 5, 4])
딸기     = np.array([5, 3, 4])
바나나   = np.array([1, 1, 5])
농구공   = np.array([4, 5, 0])

print("사과   =", 사과)
print("딸기   =", 딸기)
print("바나나 =", 바나나)
print("농구공 =", 농구공)
print()
print("👉 이렇게 '숫자 나열'로 바꾼 걸 벡터(vector)라고 불러요.")
```

## 2단계. "비슷하다"를 숫자로 재보기

이제 핵심입니다. 두 벡터가 얼마나 비슷한지 재는 가장 흔한 방법이
**코사인 유사도(cosine similarity)** 예요.

직관 딱 하나만 기억하세요:
- 두 화살표가 **같은 방향**을 가리키면 → 비슷한 것! (값이 **1에 가까움**)
- 방향이 **영 다르면** → 안 비슷한 것! (값이 **0에 가까움**)

```python
# ====================================================================
# [2번 셀] 코사인 유사도로 '비슷함' 계산하기
# ====================================================================
def 비슷한정도(a, b):
    # 코사인 유사도 = 내적 ÷ (각자 길이의 곱)
    # 수식은 몰라도 됩니다. 결과가 1에 가까우면 '비슷'!
    return np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))

print("사과 vs 딸기   :", round(비슷한정도(사과, 딸기), 3))
print("사과 vs 바나나 :", round(비슷한정도(사과, 바나나), 3))
print("사과 vs 농구공 :", round(비슷한정도(사과, 농구공), 3))
print()
print("👉 딸기가 사과랑 제일 비슷하다고 나오죠? (1에 가장 가까움)")
print("👉 방금 '비슷함'을 직접 계산한 거예요. 이게 검색·추천의 핵심입니다.")
```

## 3단계. 눈으로 보기: 비슷하면 화살표 방향이 비슷하다

숫자 3개는 그림으로 그리기 어려우니, 특징을 2개만 써서
화살표로 그려볼게요. *방향이 비슷할수록 = 비슷한 물건* 이라는 걸 눈으로 확인해봅시다.

```python
# ====================================================================
# [3번 셀] 벡터를 화살표로 그려보기
# ====================================================================
# [ 둥근정도, 단맛 ] 두 가지 특징만으로 평면에 그립니다.
물건들 = {
    '사과':   [5, 4],
    '딸기':   [3, 4],
    '바나나': [1, 5],
    '농구공': [5, 0],
}

plt.figure(figsize=(6, 6))
for 이름, v in 물건들.items():
    plt.arrow(0, 0, v[0], v[1], head_width=0.2, length_includes_head=True, color='tab:blue')
    plt.text(v[0] + 0.15, v[1] + 0.15, 이름, fontsize=14)
plt.xlim(-0.5, 6); plt.ylim(-0.5, 6)
plt.xlabel('둥근 정도'); plt.ylabel('단맛')
plt.title('화살표 방향이 비슷할수록 = 비슷한 물건')
plt.grid(True, alpha=0.3)
plt.show()
print("👉 사과와 딸기 화살표가 거의 같은 방향이죠? 그래서 '비슷'합니다.")
print("👉 농구공은 혼자 바닥(단맛 0)을 향하죠? 그래서 '안 비슷'해요.")
```

## 4단계. 진짜 문장으로! AI가 알아서 숫자로 바꿔준다

위에선 `[빨강, 둥글다, 단맛]` 숫자를 직접 정해 줬어요.
하지만 진짜 AI는 **문장을 알아서 수백 개의 숫자(벡터)로** 바꿔줍니다.
이걸 **임베딩(embedding)** 이라고 해요.

아래 셀은 한국어 AI 모델을 불러옵니다. **처음 한 번은 30초~1분** 걸려요. ☕

```python
# ====================================================================
# [4번 셀] 문장을 임베딩(숫자 벡터)으로 바꾸기
# ====================================================================
!pip install -q sentence-transformers
from sentence_transformers import SentenceTransformer

# 한국어를 잘 이해하는 작은 모델을 불러옵니다.
모델 = SentenceTransformer('jhgan/ko-sroberta-multitask')

문장들 = [
    "오늘 점심 뭐 먹지?",
    "배고픈데 밥 먹으러 갈까?",
    "내일 회의 자료 준비해야 해.",
    "프로젝트 발표 슬라이드 만들자.",
]

벡터들 = 모델.encode(문장들)
print("문장 1개가", 벡터들.shape[1], "개의 숫자로 바뀌었어요!")
print("손으로 3개 적던 걸, AI는", 벡터들.shape[1], "개를 알아서 만든 거죠.")
```

## 5단계. 방금 배운 걸로... 검색엔진 만들기 (진짜로!)

질문을 하나 던지면, 위 문장들 중 **'가장 비슷한' 걸** 찾아줍니다.
2번 셀에서 만든 `비슷한정도` 함수를 **그대로** 재사용해요.
과일에 쓰던 계산을 문장에 똑같이 쓰는 거예요.

```python
# ====================================================================
# [5번 셀] 의미로 검색하는 미니 검색엔진
# ====================================================================
질문 = "뭐 먹을까 고민돼"
질문벡터 = 모델.encode(질문)

점수들 = [비슷한정도(질문벡터, v) for v in 벡터들]
순위 = np.argsort(점수들)[::-1]  # 점수 높은 순으로 정렬

print(f"❓ 질문: {질문}")
print()
print("📊 비슷한 순서:")
for 등수, i in enumerate(순위, 1):
    print(f"  {등수}등  ({점수들[i]:.3f})  →  {문장들[i]}")
print()
print("👉 '먹을까'라는 단어가 똑같이 없는데도 밥 먹는 문장을 찾아냈죠?")
print("👉 글자가 아니라 '의미'로 검색한 겁니다. 이게 요즘 AI 검색(RAG)의 핵심이에요.")
```

## 6단계. 🚀 이제 직접 해 볼 차례: 해커톤 주제에 적용해보기

아래 `내문장들` 과 `내질문` 을 **여러분 프로젝트에 맞게** 바꿔보세요.
사내 FAQ, 제품 설명, 회의록 제목, 고객 문의... 무엇이든요.

> ⚠️ **회사 기밀·개인정보는 절대 넣지 마세요.** Colab은 외부(구글) 클라우드입니다.
> 연습은 가짜 예시 데이터로 충분해요.

```python
# ====================================================================
# [6번 셀] 내 데이터로 검색해보기 — 윗부분만 고치면 됩니다
# ====================================================================
내문장들 = [
    "여기에",
    "내 데이터를",
    "한 줄씩 적어보세요",
    "예: 비밀번호 변경은 어떻게 하나요?",
    "예: 연차 신청은 어디서 하나요?",
]
내질문 = "찾고 싶은 내용을 여기 적으세요"

# ↓↓↓ 아래는 안 바꿔도 됩니다. 그냥 실행만! ↓↓↓
내벡터들 = 모델.encode(내문장들)
내질문벡터 = 모델.encode(내질문)
내점수들 = [비슷한정도(내질문벡터, v) for v in 내벡터들]
내순위 = np.argsort(내점수들)[::-1]

print(f"❓ 질문: {내질문}")
print()
print("📊 비슷한 순서:")
for 등수, i in enumerate(내순위, 1):
    print(f"  {등수}등  ({내점수들[i]:.3f})  →  {내문장들[i]}")
```
