```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글이 네모로 깨짐)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 행렬의 열 = 기저벡터가 가는 곳: î(1,0)→첫째 열, ĵ(0,1)→둘째 열
A = np.array([[0, -1],
              [1,  0]])   # 90도 반시계 회전
print('î 가 가는 곳 (첫째 열):', A @ np.array([1, 0]))
print('ĵ 가 가는 곳 (둘째 열):', A @ np.array([0, 1]))
```

## 1. 격자 변환 그리기

변환 전(왼쪽)과 변환 후(오른쪽) 격자를 나란히 그리는 함수를 만듭니다. 초록=î, 분홍=ĵ.

```python
def plot_transform(M, title=''):
    fig, axes = plt.subplots(1, 2, figsize=(8, 4))
    for ax, T, name in [(axes[0], np.eye(2), '변환 전'), (axes[1], M, '변환 후')]:
        for i in range(-5, 6):
            v = T @ np.array([[i, i], [-5, 5]])   # 세로선 x=i
            ax.plot(v[0], v[1], color='#a5b4fc', lw=1)
            h = T @ np.array([[-5, 5], [i, i]])   # 가로선 y=i
            ax.plot(h[0], h[1], color='#a5b4fc', lw=1)
        ih, jh = T @ np.array([1, 0]), T @ np.array([0, 1])
        ax.arrow(0, 0, ih[0], ih[1], color='#10b981', width=0.06, length_includes_head=True, zorder=5)
        ax.arrow(0, 0, jh[0], jh[1], color='#ec4899', width=0.06, length_includes_head=True, zorder=5)
        ax.set_xlim(-6, 6); ax.set_ylim(-6, 6); ax.set_aspect('equal')
        ax.set_title(name)
    fig.suptitle(title)
    plt.tight_layout(); plt.show()

plot_transform(A, '90도 회전  [[0,-1],[1,0]]')
```

## 2. 대표 변환 비교

확대·전단·반사를 차례로 그려봅니다.

```python
plot_transform(np.array([[2, 0], [0, 2]]), '2배 확대  [[2,0],[0,2]]')
plot_transform(np.array([[1, 1], [0, 1]]), '전단(밀기)  [[1,1],[0,1]]')
plot_transform(np.array([[-1, 0], [0, 1]]), '좌우 반사  [[-1,0],[0,1]]')
```

## 3. 행렬 × 벡터 = 변환 적용

벡터 하나에 변환을 적용하면 행렬 곱으로 계산됩니다. 넓이 배율(det)도 확인해 봅니다.

```python
R = np.array([[0, -1], [1, 0]])   # 90도 회전
print('(1,0) 을 90도 회전 →', R @ np.array([1, 0]))   # [0 1]

shear = np.array([[1, 1], [0, 1]])
print('(2,3) 에 전단 적용 →', shear @ np.array([2, 3]))  # [5 3]

print('확대 [[2,0],[0,3]] 의 넓이 배율 det =', np.linalg.det(np.array([[2, 0], [0, 3]])))  # 6
print('납작 [[1,0],[0,0]] 의 det =', np.linalg.det(np.array([[1, 0], [0, 0]])))  # 0
```

## 4. 직접 해보기 ✏️

전단 행렬 `[[1,1],[0,1]]` 을 벡터 `(2,3)` 에 적용해 `(5,3)` 이 나오는지 확인하세요.

```python
S = np.array([[1, 1], [0, 1]])
result = S @ np.array([2, 3])
print('결과 =', result)
assert np.array_equal(result, [5, 3]), '다시 확인해 보세요!'
print('✅ 정답! (5, 3)')
```
