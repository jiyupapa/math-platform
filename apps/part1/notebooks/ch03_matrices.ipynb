## 1. 행렬 만들기: 크기(shape) 확인

`np.array` 에 리스트의 리스트를 주면 행렬이 됩니다. `.shape` 가 (행, 열).

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic'
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

A = np.array([[1, 2, 3],
              [4, 5, 6]])
print('A =\n', A)
print('크기(행, 열):', A.shape)   # (2, 3)
print('1행 2열 성분 a_12 =', A[0, 1])  # 인덱스는 0부터!
```

## 2. 덧셈과 스칼라배: 성분끼리

`+` 와 `*`(스칼라) 는 성분끼리 동작합니다.

```python
X = np.array([[1, 2], [3, 4]])
Y = np.array([[5, 6], [7, 8]])
print('X + Y =\n', X + Y)
print('2 * X =\n', 2 * X)
```

## 3. 행렬 곱셈: `@` 연산

**주의**: `*` 는 성분곱(아다마르곱), 진짜 행렬곱은 `@` 또는 `np.matmul` 입니다.
결과의 (i, j) = A의 i행 · B의 j열 (내적).

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print('A @ B (행렬곱) =\n', A @ B)
print('A * B (성분곱, 다름!) =\n', A * B)

# (2행,1열) 직접 검산: A의 2행 (3,4) 과 B의 1열 (5,7) 의 내적
print('C[1,0] =', 3*5 + 4*7)   # 43
```

## 4. 행렬 × 벡터 = 신경망 한 층 🤖

가중치 행렬 W 에 입력 벡터 x 를 곱하면 새 벡터가 나옵니다. 이것이 신경망의 핵심 연산.

```python
W = np.array([[2, 0],
              [1, 3]])
x = np.array([1, 2])

print('W @ x =', W @ x)   # [2, 7]
# 1번째: 2*1 + 0*2 = 2,  2번째: 1*1 + 3*2 = 7
```

## 5. 이미지 = 숫자 행렬 🖼️

컴퓨터에게 이미지는 픽셀 밝기를 담은 행렬일 뿐입니다. 숫자 "1" 모양을 그려봅시다.

```python
img = np.array([
    [0, 0, 7, 0, 0],
    [0, 7, 9, 0, 0],
    [0, 0, 9, 0, 0],
    [0, 0, 9, 0, 0],
    [0, 7, 9, 7, 0],
])
print('이미지 행렬의 크기:', img.shape)

plt.figure(figsize=(3, 3))
plt.imshow(img, cmap='gray')
plt.title('같은 행렬을 그림으로')
plt.axis('off')
plt.show()
```

## 6. 직접 풀어보기 ✏️

`A @ B` 의 (1행, 1열) 값을 손으로 예측한 뒤, 코드로 확인하세요. (정답: 1×5 + 2×7 = 19)

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
C = A @ B
print('C[0,0] =', C[0, 0])
assert C[0, 0] == 19, '다시 계산해 보세요!'
print('✅ 정답! 1×5 + 2×7 = 19')
```
