## 1. 두 직선의 교점 = 해

$$\begin{cases} x + y = 3 \\ x - y = 1 \end{cases}$$

두 직선을 그리면 만나는 점이 해입니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글이 네모로 깨짐)
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

x = np.linspace(-2, 6, 100)
y1 = 3 - x       # x + y = 3  ->  y = 3 - x
y2 = x - 1       # x - y = 1  ->  y = x - 1

plt.figure(figsize=(5, 5))
plt.plot(x, y1, label='x + y = 3', color='#6366f1')
plt.plot(x, y2, label='x - y = 1', color='#ec4899')
plt.plot(2, 1, 'o', color='#10b981', markersize=10, label='교점 (2, 1)')
plt.axhline(0, color='#94a3b8', lw=1); plt.axvline(0, color='#94a3b8', lw=1)
plt.grid(True, color='#eef2f7'); plt.gca().set_aspect('equal')
plt.legend(); plt.title('두 직선의 교점 = 연립방정식의 해')
plt.show()
```

## 2. 행렬로 풀기 — A x = b

계수행렬 A 와 상수벡터 b 를 만들고 `np.linalg.solve(A, b)` 로 한 번에 풉니다.

```python
A = np.array([[1, 1],
              [1, -1]])
b = np.array([3, 1])

sol = np.linalg.solve(A, b)
print('해 (x, y) =', sol)   # [2. 1.]
```

## 3. 가우스 소거법 직접 해보기

첨가행렬에 행 연산을 적용해 손으로 푸는 과정을 코드로 따라가 봅시다.
(R2 ← R2 − R1 로 x를 소거)

```python
M = np.array([[1.0, 1.0, 3.0],
              [1.0, -1.0, 1.0]])   # 첨가행렬 [A | b]
print('시작:\n', M)

M[1] = M[1] - M[0]    # R2 <- R2 - R1 (둘째 식에서 x 소거)
print('\nR2 <- R2 - R1:\n', M)

y = M[1, 2] / M[1, 1]          # -2y = -2 -> y = 1
x = (M[0, 2] - M[0, 1] * y) / M[0, 0]   # x + y = 3 -> x = 2
print(f'\n후진대입 결과: x = {x}, y = {y}')
```

## 4. 해의 종류 — 행렬식으로 판별 🤖

계수행렬의 **행렬식(det)** 이 0이 아니면 유일해, 0이면 (해 없음 또는 무수히 많음).
행렬식은 다음 챕터에서 자세히 배웁니다.

```python
uniq = np.array([[1, 1], [1, -1]])   # 교차
para = np.array([[1, 1], [1, 1]])    # 평행/일치 (det=0)
print('유일해 케이스 det =', round(np.linalg.det(uniq), 2))   # != 0
print('det=0 케이스 det =', round(np.linalg.det(para), 2))    # 0
```

## 5. 직접 풀어보기 ✏️

`2x + y = 5`, `x − y = 1` 을 `np.linalg.solve` 로 풀어보세요. (정답: x=2, y=1)

```python
A = np.array([[2, 1], [1, -1]])
b = np.array([5, 1])
sol = np.linalg.solve(A, b)
print('해 =', sol)
assert np.allclose(sol, [2, 1]), '다시 확인해 보세요!'
print('✅ 정답! x=2, y=1')
```
