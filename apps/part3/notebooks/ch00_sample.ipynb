## 0. 준비 — 라이브러리 불러오기

`numpy` 는 숫자 계산, `matplotlib` 는 그래프 도구입니다. 몰라도 괜찮아요 —
셀을 실행(`Shift + Enter`)하고 결과만 눈으로 보면 됩니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (Windows) — 그래프의 한글이 깨지지 않게
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

# 난수 생성기 — 시드(42)를 고정하면 실행할 때마다 같은 결과가 나와요
rng = np.random.default_rng(42)
print('준비 완료!')
```

## 1. 동전 열 번 던지기

`rng.integers(0, 2, size=10)` 은 0(뒤) 또는 1(앞)을 10개 뽑습니다.
앞면 확률이 0.5 라고 해서 10번 중 정확히 5번 나올까요?

```python
flips = rng.integers(0, 2, size=10)   # 1 = 앞면, 0 = 뒷면
print('던진 결과 :', flips)
print('앞면 횟수 :', flips.sum(), '/ 10')
print('앞면 비율 :', flips.sum() / 10)
```

10번으로는 0.5 에서 꽤 벗어나는 게 보통입니다. 그럼 **2000번** 던지면서
매 순간의 누적 비율을 추적해 볼까요?

## 2. 던질수록 0.5 로 — 비율 수렴 그래프

```python
n = 2000
flips = rng.integers(0, 2, size=n)
# 누적 비율: k번째까지의 앞면 수 ÷ k  (np.cumsum = 누적 합)
ratio = np.cumsum(flips) / np.arange(1, n + 1)

plt.figure(figsize=(8, 4))
plt.plot(ratio, color='#6366f1', label='누적 앞면 비율')
plt.axhline(0.5, color='#f59e0b', linestyle='--', label='이론값 0.5')
plt.xlabel('던진 횟수')
plt.ylabel('앞면 비율')
plt.ylim(0, 1)
plt.title('동전을 던질수록 앞면 비율은 0.5 로 (큰 수의 법칙 맛보기)')
plt.legend()
plt.show()

print('처음 10번의 비율 :', round(ratio[9], 3))
print('2000번의 비율   :', round(ratio[-1], 3))
```

그래프 왼쪽(적은 횟수)에서는 크게 출렁이다가, 오른쪽으로 갈수록 0.5 근처에서
잠잠해지는 게 보이죠? 이것이 웹 챕터에서 본 **큰 수의 법칙**의 맛보기입니다 (7장에서 제대로!).

🧪 **직접 실험**: 위 셀에서 `n = 2000` 을 `n = 100` 이나 `n = 20000` 으로 바꾸고
다시 실행해 보세요. 시드도 `rng = np.random.default_rng(7)` 처럼 바꾸면 다른 우연이 펼쳐집니다.

## 3. 주사위 확률 — 경우의 수 vs 시뮬레이션

주사위에서 3의 배수(3, 6)가 나올 확률은 경우의 수로 2/6 = 1/3 ≈ 0.333.
정말 그런지 10만 번 던져서 확인해 봅시다.

```python
rolls = rng.integers(1, 7, size=100_000)   # 1~6 의 눈을 10만 번
multiple_of_3 = (rolls % 3 == 0)           # 3의 배수인가? (True/False 배열)

print('이론값   : 2/6 =', round(2 / 6, 4))
print('시뮬레이션:', multiple_of_3.mean())   # True 비율 = 경험적 확률
```

## 정리

- **확률 = 0~1 사이의 숫자**이고, 그 진짜 뜻은 "오래 반복했을 때 다가가는 **비율**".
- 적은 횟수에서는 들쭉날쭉이 정상 — 횟수를 늘릴수록 이론값에 수렴합니다.
- 경우의 수로 구한 이론 확률(1/3)과 시뮬레이션 비율이 거의 일치했어요.
