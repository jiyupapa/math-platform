## 1. 편미분: 한 변수만, 나머지는 상수

SymPy 의 `sp.diff(f, 변수)` 로 편미분합니다. 나머지 변수는 자동으로 상수 취급돼요.

이 편미분은 ∂f/∂x("라운드 에프 라운드 엑스")로 씁니다. ∂(라운드 디/파셜)는 변수가 여럿일 때 미분에 쓰는 기호예요.

```python
import sympy as sp

x, y = sp.symbols('x y')
f = x**2 + y**2

print('∂f/∂x =', sp.diff(f, x))   # 2x  (y² 은 상수 → 0)
print('∂f/∂y =', sp.diff(f, y))   # 2y
print('점 (3,2): ∂f/∂x =', sp.diff(f, x).subs({x: 3, y: 2}),
      ', ∂f/∂y =', sp.diff(f, y).subs({x: 3, y: 2}))
```

## 2. 다른 예: xy, x²y

다른 변수를 상수로 보는 감각을 익힙니다.

```python
print('f=xy   → ∂/∂x =', sp.diff(x*y, x), ', ∂/∂y =', sp.diff(x*y, y))       # y , x
print('f=x²y  → ∂/∂x =', sp.diff(x**2*y, x), ', ∂/∂y =', sp.diff(x**2*y, y))  # 2xy , x²
print('x²y 의 점 (2,3): ∂/∂x =', sp.diff(x**2*y, x).subs({x: 2, y: 3}))        # 12
```

## 3. 곡면과 두 방향 단면

f(x,y)=x²+y² 은 밥그릇 모양 곡면. 점 (1,1) 을 지나는 두 방향 단면 곡선의 기울기가
각각 ∂f/∂x, ∂f/∂y 입니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic' 로 바꾸세요
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호(-) 깨짐 방지

def F(x, y):
    return x**2 + y**2

gx = np.linspace(-3, 3, 60)
gy = np.linspace(-3, 3, 60)
X, Y = np.meshgrid(gx, gy)
Z = F(X, Y)

fig = plt.figure(figsize=(11, 4))

# 3D 곡면
ax0 = fig.add_subplot(1, 3, 1, projection='3d')
ax0.plot_surface(X, Y, Z, cmap='Blues', alpha=0.8)
ax0.scatter([1], [1], [F(1, 1)], color='red', s=40)
ax0.set_title('f(x,y)=x²+y²')

# x 방향 단면 (y=1 고정): g(x)=x²+1, 기울기 2x → x=1 이면 2
ax1 = fig.add_subplot(1, 3, 2)
ax1.plot(gx, F(gx, 1), color='#f59e0b', lw=2.5)
ax1.plot(gx, F(1, 1) + 2*1*(gx - 1), color='#ef4444', lw=1.8)   # 접선 기울기 ∂f/∂x=2
ax1.scatter([1], [F(1, 1)], color='#ef4444'); ax1.grid(True, color='#eef2f7')
ax1.set_title('x 방향 단면 (∂f/∂x=2)')

# y 방향 단면 (x=1 고정): h(y)=1+y², 기울기 2y → y=1 이면 2
ax2 = fig.add_subplot(1, 3, 3)
ax2.plot(gy, F(1, gy), color='#10b981', lw=2.5)
ax2.plot(gy, F(1, 1) + 2*1*(gy - 1), color='#ef4444', lw=1.8)   # 접선 기울기 ∂f/∂y=2
ax2.scatter([1], [F(1, 1)], color='#ef4444'); ax2.grid(True, color='#eef2f7')
ax2.set_title('y 방향 단면 (∂f/∂y=2)')

plt.tight_layout(); plt.show()
```
