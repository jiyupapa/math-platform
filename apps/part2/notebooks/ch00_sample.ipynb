## 1. 함수 만들기: f(x) = x²

미분의 주인공인 함수를 파이썬 함수로 정의합니다.

```python
def f(x):
    return x**2

print('f(3) =', f(3))   # 높이(함숫값)
print('f(2) =', f(2))
```

## 2. 평균변화율 (할선의 기울기)

두 점 $x$ 와 $x+h$ 를 잇는 할선의 기울기:

$$\frac{f(x+h) - f(x)}{h}$$

웹의 **노란 할선**이 바로 이것입니다. h 를 점점 줄여봅시다.

```python
def average_rate(x, h):
    return (f(x + h) - f(x)) / h

x = 2
for h in [1, 0.5, 0.1, 0.01, 0.001]:
    print(f'h = {h:<6}  평균변화율 = {average_rate(x, h):.4f}')
```

h 가 0 에 가까워질수록 값이 **4** 에 다가갑니다.
이게 바로 x=2 에서의 **순간변화율(접선 기울기) = f'(2)** 입니다.
f'(x) = 2x 이므로 2 × 2 = 4 와 정확히 일치하죠.

(`f'` 는 "f 프라임"이라 읽고, **미분한 값 = 기울기**를 뜻하는 표기예요.)

## 3. 곡선과 접선 그리기

Matplotlib 으로 f(x)=x² 곡선과, x=1 에서의 접선을 함께 그립니다.
웹 캔버스에서 본 그 빨간 접선과 같은 것입니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic' 로 바꾸세요
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호(-) 깨짐 방지

x0 = 1.0
slope = 2 * x0            # f'(x) = 2x
xs = np.linspace(-3, 3, 200)

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(xs, f(xs), color='#6366f1', lw=2.5, label='f(x) = x^2')
ax.plot(xs, f(x0) + slope * (xs - x0), color='#ef4444', lw=2,
        label=f'x={x0} 에서의 접선 (기울기 {slope:.0f})')
ax.scatter([x0], [f(x0)], color='#ef4444', zorder=5)

ax.axhline(0, color='#94a3b8', lw=1); ax.axvline(0, color='#94a3b8', lw=1)
ax.set_xlim(-3, 3); ax.set_ylim(-1, 9)
ax.grid(True, color='#eef2f7'); ax.legend()
ax.set_title('곡선과 접선')
plt.show()
```

## 4. SymPy 로 기호 미분하기 ✨

`sympy` 는 수식을 **기호 그대로** 다뤄서, 사람이 손으로 하듯 미분 공식을 적용해 줍니다.
직접 미분한 결과를 검산할 때 아주 유용합니다.

```python
import sympy as sp

x = sp.Symbol('x')
expr = x**2
deriv = sp.diff(expr, x)      # x^2 를 x 로 미분

print('f(x)  =', expr)
print("f'(x) =", deriv)        # -> 2*x
print("x=3 에서 f'(3) =", deriv.subs(x, 3))   # -> 6
```
