## 1. 거듭제곱 미분 공식: (x^n)' = n·x^(n-1)

지수를 1 부터 5 까지 바꿔가며 미분 결과의 패턴을 확인합니다.
지수가 앞으로 내려와 계수가 되고, 지수는 1 줄어듭니다.

```python
import sympy as sp

x = sp.Symbol('x')
for n in range(1, 6):
    print(f'(x^{n})\' =', sp.diff(x**n, x))
```

## 2. 상수 · 상수배 · 합차

- 상수의 미분 = 0
- 상수배: 계수는 그대로 앞에
- 합·차: 항마다 따로 미분해 더함

```python
print('(7)\'       =', sp.diff(7, x))            # 0
print('(3x^2)\'    =', sp.diff(3*x**2, x))       # 6x
print('(5x^4)\'    =', sp.diff(5*x**4, x))       # 20x^3
print('(x^2+3x)\'  =', sp.diff(x**2 + 3*x, x))   # 2x+3
print('(x^4-2x)\'  =', sp.diff(x**4 - 2*x, x))   # 4x^3-2
```

## 3. 다항식 통째로 미분 + 손 계산 검산

f(x) = x³ + 2x² + 5x + 7 을 미분하면 3x² + 4x + 5.
손으로 항별 미분한 것과 SymPy 결과가 같은지 확인합니다.

```python
f = x**3 + 2*x**2 + 5*x + 7
by_sympy = sp.diff(f, x)
by_hand = 3*x**2 + 4*x + 5     # 항별로 손 미분

print('SymPy :', by_sympy)
print('손 계산:', by_hand)
print('같은가?:', sp.simplify(by_sympy - by_hand) == 0)
print("f'(1) =", by_sympy.subs(x, 1))   # 3+4+5 = 12
```

## 4. f 와 f' 그래프

다항식과 그 도함수를 나란히 그려봅니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic' 로 바꾸세요
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호(-) 깨짐 방지

# SymPy 식을 빠른 수치 함수로 변환
f_np = sp.lambdify(x, f, 'numpy')
df_np = sp.lambdify(x, by_sympy, 'numpy')

xs = np.linspace(-3, 2, 200)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
ax1.plot(xs, f_np(xs), color='#6366f1', lw=2.5); ax1.set_title('f(x) = x³+2x²+5x+7')
ax2.plot(xs, df_np(xs), color='#10b981', lw=2.5); ax2.set_title("f'(x) = 3x²+4x+5")
for ax in (ax1, ax2):
    ax.axhline(0, color='#94a3b8', lw=1); ax.axvline(0, color='#94a3b8', lw=1)
    ax.grid(True, color='#eef2f7')
plt.tight_layout(); plt.show()
```

## 5. 직접 풀어보기 ✏️

f(x)=x³ 을 거듭제곱 공식으로 미분해 **f'(2)** 를 구해 보세요. (웹 문제와 같습니다. 답은 12!)

```python
# TODO: (x³)' = 3x² 이므로 x=2 를 대입해 answer 에 담으세요.
# 힌트: sp.diff(x**3, x).subs(x, 2)

answer = sp.diff(x**3, x).subs(x, 2)
print('정답:', answer)
assert answer == 12, '다시! (x³)′ = 3x², x=2 이면 3×4.'
print('✅ 잘했어요! (x³)′=3x², f\'(2)=12.')
```
