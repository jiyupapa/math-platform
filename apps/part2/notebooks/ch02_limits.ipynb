## 1. 평균변화율 (차분몫)

$$\text{평균변화율} = \frac{f(x+h) - f(x)}{h}$$

구간 x 부터 x+h 까지의 기울기(할선 기울기)입니다.

```python
def f(x):
    return x**2

def average_rate(x, h):
    return (f(x + h) - f(x)) / h

# x=1 부터 x=3 까지의 평균변화율
print('x=1~3 평균변화율:', average_rate(1, 2))   # (9-1)/2 = 4
```

## 2. h 를 줄이면? 극한 표

x=1 에서 h 를 양쪽(＋, －)에서 0 으로 보내며 평균변화율을 봅니다.
차분몫이 (2 + h) 로 정리되므로 **2** 에 수렴해야 합니다.

> **왜 극한을 쓰냐면:** h=0 을 그냥 넣으면 분모가 0 이라 (f(1)−f(1))/0 = 0/0 이 되어 계산할 수 없어요.
> 그래서 "h 를 0 에 가까이 보낼 때 다가가는 값"을 봅니다.
> (차분몫 펼치기는 중학교 **분배법칙** 그대로: (1+h)(1+h) = 1·1 + 1·h + h·1 + h·h = 1 + 2h + h²,
> 가운데 두 항이 같아서 합치면 2h. 그래서 ((1+h)²−1)/h = (2h+h²)/h = **2+h**.)

```python
print(f"{'|h|':>8} | {'왼쪽 h<0':>12} | {'오른쪽 h>0':>12}")
print('-' * 40)
for m in [1, 0.5, 0.1, 0.01, 0.001]:
    print(f'{m:>8} | {average_rate(1, -m):>12.4f} | {average_rate(1, m):>12.4f}')
print('-' * 40)
print('h → 0 이면 양쪽 모두 → 2 로 수렴')
```

## 3. 할선이 접선으로 (그래프)

h 를 줄이면 할선(회색)이 접선(빨강)에 점점 붙습니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic' 로 바꾸세요
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호(-) 깨짐 방지

a = 1.0
xs = np.linspace(-1, 3, 200)

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(xs, f(xs), color='#6366f1', lw=2.5, label='f(x) = x^2')

# 할선들 (h 가 클수록 옅게)
for h, alpha in [(1.5, 0.35), (1.0, 0.55), (0.5, 0.8)]:
    m = average_rate(a, h)
    ax.plot(xs, f(a) + m * (xs - a), color='#f59e0b', lw=1.5, alpha=alpha,
            label=f'할선 h={h} (기울기 {m:.2f})')

# 접선 (h→0 극한, 기울기 2)
ax.plot(xs, f(a) + 2 * (xs - a), color='#ef4444', lw=2.5, label='접선 (기울기 2)')
ax.scatter([a], [f(a)], color='#ef4444', zorder=5)

ax.axhline(0, color='#94a3b8', lw=1); ax.axvline(0, color='#94a3b8', lw=1)
ax.set_xlim(-1, 3); ax.set_ylim(-1, 9)
ax.grid(True, color='#eef2f7'); ax.legend(fontsize=8)
ax.set_title('할선이 접선으로 (h → 0)')
plt.show()
```

## 4. SymPy 로 극한 계산하기 ✨

`sp.limit` 으로 차분몫의 h→0 극한을 **기호 그대로** 구합니다.
이것이 곧 순간변화율(= 도함수)입니다.

```python
import sympy as sp

x, h = sp.symbols('x h')
expr = x**2

avg = (expr.subs(x, x + h) - expr) / h
print('차분몫 정리:', sp.simplify(avg))          # -> 2*x + h

inst = sp.limit(avg, h, 0)
print('순간변화율 lim(h→0):', inst)               # -> 2*x
print('x=1 에서:', inst.subs(x, 1))               # -> 2
print('x=3 에서:', inst.subs(x, 3))               # -> 6
```

## 5. 직접 풀어보기 ✏️

```python
# TODO: x=3 에서의 순간변화율을 구해 answer 에 담으세요.
# 힌트: inst.subs(x, 3)  또는  sp.limit(average_rate 식, h, 0)

answer = sp.limit((f(3 + h) - f(3)) / h, h, 0)
print('정답:', answer)
assert answer == 6, '다시! (3+h)² − 9 = 6h + h², 이걸 h 로 나누면 6 + h → 6.'
print('✅ 잘했어요! x=3 에서 x² 의 순간변화율은 6 (= 2×3).')
```
