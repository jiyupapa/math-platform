## 1. 함수는 '입력 → 출력' 기계

f(x) = 2x + 1 을 파이썬 함수로 정의합니다. 입력을 넣으면 출력이 나옵니다.

```python
def f(x):
    return 2 * x + 1

for x in [-2, 0, 3]:
    print(f'f({x}) = {f(x)}')
```

## 2. 그래프로 보기

여러 입력에 대한 출력을 좌표평면에 찍어 이으면 **직선**이 됩니다.
그래프는 곧 점 (x, f(x)) 들의 모음입니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic' 로 바꾸세요
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호(-) 깨짐 방지

xs = np.linspace(-3, 3, 100)

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(xs, f(xs), color='#6366f1', lw=2.5, label='f(x) = 2x + 1')
ax.scatter([0], [f(0)], color='#6366f1', zorder=5)
ax.annotate('y절편 (0, 1)', (0, f(0)), textcoords='offset points', xytext=(8, -4), color='#6366f1')

ax.axhline(0, color='#94a3b8', lw=1); ax.axvline(0, color='#94a3b8', lw=1)
ax.set_xlim(-5, 5); ax.set_ylim(-5, 8)
ax.grid(True, color='#eef2f7'); ax.legend()
ax.set_title('일차함수의 그래프는 직선')
plt.show()
```

## 3. 두 점으로 기울기 구하기

$$\text{기울기} = \frac{\Delta y}{\Delta x} = \frac{y_2 - y_1}{x_2 - x_1}$$

직선 위 아무 두 점을 골라도 기울기는 항상 같습니다 (직선의 특징).

(Δ 는 "델타", 변화량(얼마나 변했나)을 뜻해요. Δy=세로 변화, Δx=가로 변화.)

```python
def slope(p1, p2):
    (x1, y1), (x2, y2) = p1, p2
    return (y2 - y1) / (x2 - x1)

# f(x)=2x+1 위의 세 쌍 — 어떤 쌍을 골라도 기울기는 2
print('기울기 (0,1)-(2,5):', slope((0, f(0)), (2, f(2))))
print('기울기 (1,3)-(4,9):', slope((1, f(1)), (4, f(4))))
print('기울기 (-2,-3)-(3,7):', slope((-2, f(-2)), (3, f(3))))
```

## 4. SymPy 로 확인: 일차함수를 미분하면 기울기가 나온다 ✨

미분은 '기울기를 구하는 도구'입니다. 일차함수는 어디서나 기울기가 같으므로,
미분하면 **상수(기울기 a)** 하나가 나옵니다. 다음 챕터의 예고편이에요.

```python
import sympy as sp

x = sp.Symbol('x')
expr = 2 * x + 1
print('f(x)  =', expr)
print("f'(x) =", sp.diff(expr, x))   # -> 2  (기울기)

# 기울기가 a 인 일반 일차함수도 미분하면 a 가 나온다
a, b = sp.symbols('a b')
print("d/dx (a*x + b) =", sp.diff(a * x + b, x))   # -> a
```
