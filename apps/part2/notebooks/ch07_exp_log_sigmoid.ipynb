## 1. 지수함수와 자연로그

- (eˣ)' = eˣ  (미분해도 자기 자신)
- (ln x)' = 1/x

```python
import sympy as sp

x = sp.Symbol('x', positive=True)   # ln 을 위해 x>0
print('(e^x)\'  =', sp.diff(sp.exp(x), x))    # exp(x)
print('(ln x)\' =', sp.diff(sp.ln(x), x))     # 1/x
print('(ln x)\' at x=4 :', sp.diff(sp.ln(x), x).subs(x, 4))   # 1/4
```

## 2. 시그모이드와 그 미분: σ' = σ(1−σ)

SymPy 로 시그모이드를 미분하고, σ(1−σ) 와 정말 같은지 확인합니다.

시그모이드는 σ(t) = 1/(1+e^{-t}) 예요 (어떤 실수든 0~1 로 눌러주는 S자 함수). e^{-t} 는 e^t 의 역수라 t 가 커질수록 0 에 가까워집니다.

```python
t = sp.Symbol('t')
sig = 1 / (1 + sp.exp(-t))
dsig = sp.diff(sig, t)

print('σ(t)  =', sig)
print('σ\'(t) =', sp.simplify(dsig))
# σ(1-σ) 와 같은가?
print('σ(1−σ) 와 같은가?:', sp.simplify(dsig - sig * (1 - sig)) == 0)
print('σ\'(0) =', dsig.subs(t, 0))    # 1/4 = 0.25
```

## 3. 그래프: S자와 종 모양, 그리고 기울기 소실

σ 는 S자, σ' 는 종 모양(최대 0.25). 양 끝에서 σ' 가 0 에 가까워지는 것이 기울기 소실입니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic' 로 바꾸세요
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호(-) 깨짐 방지

def sigmoid(x):
    return 1 / (1 + np.exp(-x))

xs = np.linspace(-6, 6, 300)
s = sigmoid(xs)
ds = s * (1 - s)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4))
ax1.plot(xs, s, color='#6366f1', lw=2.5); ax1.set_title('σ(x): S자 (0~1)')
ax1.axhline(0.5, color='#cbd5e1', ls='--', lw=1)
ax2.plot(xs, ds, color='#ef4444', lw=2.5); ax2.set_title("σ'(x)=σ(1−σ), 최대 0.25")
ax2.axhline(0.25, color='#fca5a5', ls='--', lw=1)
for ax in (ax1, ax2):
    ax.axvline(0, color='#94a3b8', lw=1); ax.grid(True, color='#eef2f7')
plt.tight_layout(); plt.show()

print('양 끝 기울기(소실):  σ\'(-6)=%.5f,  σ\'(6)=%.5f' % (float(sigmoid(-6)*(1-sigmoid(-6))), float(sigmoid(6)*(1-sigmoid(6)))))
```

## 4. 연쇄법칙과 함께 (속미분 잊지 않기)

(e^{2x})' = 2e^{2x},  (e^{3x})' = 3e^{3x}. 속미분이 앞에 곱해집니다.

```python
print('(e^{2x})\' =', sp.diff(sp.exp(2*x), x))     # 2*exp(2x)
print('(e^{3x})\' =', sp.diff(sp.exp(3*x), x))     # 3*exp(3x)
print('(e^{2x})\' at x=0 :', sp.diff(sp.exp(2*x), x).subs(x, 0))   # 2
```

## 5. 직접 풀어보기 ✏️

시그모이드의 **σ'(0)** 을 구해 보세요. (웹 문제와 같습니다. 답은 0.25!)

```python
# TODO: σ'(0) 을 구해 answer 에 담으세요.
# 힌트: σ(0)=0.5 이므로 σ(1−σ)=0.5*0.5, 또는 float(dsig.subs(t, 0))

answer = float(dsig.subs(t, 0))
print('정답:', answer)
assert abs(answer - 0.25) < 1e-9, '다시! σ(0)=0.5 이므로 0.5×0.5=0.25.'
print('✅ 잘했어요! σ\'(0)=0.25, 시그모이드 기울기의 최댓값이에요.')
```
