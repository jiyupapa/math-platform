## 1. 곱의 함정: (fg)' ≠ f'g'

각각 미분해서 곱하면 틀립니다. x·x 로 확인해 봅시다.

```python
import sympy as sp

x = sp.Symbol('x')

wrong = sp.diff(x, x) * sp.diff(x, x)   # f'·g' = 1·1 = 1  (틀린 방법)
right = sp.diff(x * x, x)               # (x·x)' = 2x       (맞는 답)
print('틀린 방법 f\'·g\' :', wrong)
print('맞는 답 (x·x)\'   :', right)
```

## 2. 곱의 법칙: (fg)' = f'g + fg'

손으로 적용한 곱의 법칙과 SymPy 결과가 같은지 확인합니다.

```python
f = x
g = x + 1

by_rule = sp.diff(f, x) * g + f * sp.diff(g, x)   # f'g + fg'
by_sympy = sp.diff(f * g, x)

print('곱의 법칙(손) :', sp.expand(by_rule))
print('SymPy        :', sp.expand(by_sympy))
print('같은가?       :', sp.simplify(by_rule - by_sympy) == 0)
```

## 3. 몫의 법칙: (f/g)' = (f'g − fg') / g²

분자의 빼는 순서, 분모의 제곱에 주의하세요.

> 왜 이런 모양이냐면, f/g = f × (1/g) 로 바꿔 놓고 방금 배운 **곱의 법칙**을 적용하면
> 이 공식이 저절로 나옵니다(1/g 의 미분이 −g′/g² 이라 부호와 분모 제곱이 생겨요).

```python
num = sp.diff(f, x) * g - f * sp.diff(g, x)   # f'g - fg'
by_rule_q = num / g**2
by_sympy_q = sp.diff(f / g, x)

print('몫의 법칙(손) :', sp.simplify(by_rule_q))
print('SymPy        :', sp.simplify(by_sympy_q))
print('같은가?       :', sp.simplify(by_rule_q - by_sympy_q) == 0)
print('x=1 에서 값   :', by_sympy_q.subs(x, 1))   # 1/4
```

## 4. 곱의 법칙을 넓이로 보기 (수치 확인)

직사각형 넓이 u·v 가 x 를 Δ 만큼 늘릴 때 얼마나 변하는지 쪼개봅니다.
Δ 를 줄이면 (넓이 변화 ÷ Δ) 가 곱의 법칙 값에 수렴합니다.

```python
def u(t):
    return t
def v(t):
    return t + 1

x0 = 1.0
exact = 1 * v(x0) + u(x0) * 1        # u'v + uv' (곱의 법칙, u'=v'=1)
print('곱의 법칙 정확값 u\'v+uv\':', exact)
for d in [0.5, 0.1, 0.01, 0.001]:
    change = u(x0 + d) * v(x0 + d) - u(x0) * v(x0)
    print(f'Δ={d:<6}  넓이변화/Δ = {change / d:.4f}')
```

## 5. 직접 풀어보기 ✏️

곱의 법칙으로 **(x·(x+1))' 을 x=2 에서** 구해 보세요. (웹 문제와 같습니다. 답은 5!)

```python
# TODO: (x·(x+1))' 을 구해 x=2 를 대입해 answer 에 담으세요.
# 힌트: sp.diff(x*(x+1), x).subs(x, 2)

answer = sp.diff(x * (x + 1), x).subs(x, 2)
print('정답:', answer)
assert answer == 5, '다시! (x(x+1))′ = 2x+1, x=2 이면 5.'
print('✅ 잘했어요! (x(x+1))′ = 2x+1, x=2 → 5.')
```
