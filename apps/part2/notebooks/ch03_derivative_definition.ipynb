## 1. 미분계수: 정의(극한)로 근사하기

$$f'(a) = \lim_{h\to 0}\frac{f(a+h)-f(a)}{h}$$

아주 작은 h 를 넣어 x=2 에서의 미분계수 f'(2) 를 근사해 봅니다. (참값은 4)

```python
def f(x):
    return x**2

def deriv_at(a, h):
    return (f(a + h) - f(a)) / h

for h in [0.1, 0.01, 0.001, 0.0001]:
    print(f'h={h:<7}  f\'(2) 근사 = {deriv_at(2, h):.5f}')
print('→ 4 로 수렴 (미분계수 f\'(2) = 4)')
```

## 2. 도함수는 '함수': f 와 f' 를 나란히

각 점의 기울기(미분계수)를 모으면 도함수 f'(x)=2x 가 됩니다.
왼쪽 f(x)=x² 의 기울기가 오른쪽 f'(x) 의 높이로 나타납니다.

```python
import numpy as np
import matplotlib.pyplot as plt

# 한글 폰트 설정 (안 하면 한글 제목/라벨이 네모로 깨짐)
# Windows: 'Malgun Gothic' / macOS: 'AppleGothic' / Linux: 'NanumGothic' 로 바꾸세요
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False  # 음수 부호(-) 깨짐 방지

def df(x):
    return 2 * x   # 도함수

xs = np.linspace(-3, 3, 200)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4.5))

ax1.plot(xs, f(xs), color='#6366f1', lw=2.5)
ax1.set_title('f(x) = x^2 (곡선)')

ax2.plot(xs, df(xs), color='#10b981', lw=2.5)
ax2.set_title("f'(x) = 2x (도함수)")

for ax in (ax1, ax2):
    ax.axhline(0, color='#94a3b8', lw=1); ax.axvline(0, color='#94a3b8', lw=1)
    ax.grid(True, color='#eef2f7')

# x=1 에서: 왼쪽 접선 기울기 2 = 오른쪽 점 높이 2
a = 1
ax1.plot(xs, f(a) + df(a) * (xs - a), color='#ef4444', lw=1.8)
ax1.scatter([a], [f(a)], color='#ef4444', zorder=5)
ax2.scatter([a], [df(a)], color='#ef4444', zorder=5)
ax2.annotate(f"f'({a})={df(a)}", (a, df(a)), textcoords='offset points', xytext=(6, -12), color='#ef4444')

plt.tight_layout(); plt.show()
```

## 3. SymPy 로 정의부터 도함수까지 ✨

차분몫의 극한(정의)으로 도함수를 구하고, `sp.diff` 결과와 같은지 확인합니다.
그런 다음 특정 점에 대입해 **미분계수**를 얻습니다.

```python
import sympy as sp

x, h = sp.symbols('x h')
expr = x**2

# 정의(극한)로 도함수
by_def = sp.limit((expr.subs(x, x + h) - expr) / h, h, 0)
print('정의로 구한 도함수 :', by_def)          # 2*x

# 공식(diff)로 도함수
by_diff = sp.diff(expr, x)
print('sp.diff 도함수     :', by_diff)          # 2*x
print('둘이 같은가?        :', sp.simplify(by_def - by_diff) == 0)

# 도함수에 대입 → 미분계수
print("미분계수 f'(2)      :", by_diff.subs(x, 2))   # 4
print("미분계수 f'(5)      :", by_diff.subs(x, 5))   # 10
```

```python
# TODO: 도함수에 x=5 를 대입해 answer 에 담으세요.
# 힌트: sp.diff(x**2, x).subs(x, 5)  또는  2*5

answer = sp.diff(x**2, x).subs(x, 5)
print('정답:', answer)
assert answer == 10, "다시! 도함수 f'(x)=2x 에 x=5 를 넣습니다."
print("✅ 잘했어요! f'(5)=10. 도함수 하나로 어느 점이든 대입만 하면 끝.")
```
