## 검산대 ① 예산·배분 (1·4장): 스펙이 서로 맞물리나

```python
import math

N_PARAM, D_TOK = 40e9, 1.2e12
C = 6 * N_PARAM * D_TOK
gpu_days = C / (312e12 * 0.5) / 86400
per_4000 = gpu_days / 4000
ratio = D_TOK / N_PARAM

print(f'C = 6·N·D = {C:.2e} FLOPs')
print(f'GPU-일 = {gpu_days:,.0f} → 4,000장이면 {per_4000:.1f}일  (주장: 약 5일)')
print(f'토큰/파라미터 = {ratio:.0f} : 1  (황금비 20 근처, 서빙 약간 고려한 배분)')

assert abs(C - 2.88e23) < 1e15
assert round(per_4000, 1) == 5.3, '"약 5일" 주장과 일치 — 검산 통과 ✓'
assert ratio == 30
print()
print('✅ 판정: 예산·배분 주장은 검산과 일치. 유일한 통과')
```

## 검산대 ② "0.8점 차 1위" (6·7장): 오차범위 안인가

```python
def half_width(p, n):
    return 1.96 * math.sqrt(p * (1 - p) / n) * 100

n_bench = 2000
terra, nuri = 71.2, 70.4
hw_t = half_width(terra / 100, n_bench)
hw_n = half_width(nuri / 100, n_bench)
print(f'테라-1: {terra} ± {hw_t:.1f} → [{terra - hw_t:.1f}, {terra + hw_t:.1f}]')
print(f'누리-7: {nuri} ± {hw_n:.1f} → [{nuri - hw_n:.1f}, {nuri + hw_n:.1f}]')

overlap = terra - hw_t < nuri + hw_n
assert round(hw_t, 1) == 2.0 and overlap, '0.8점 차이는 ±2.0 구간이 크게 겹친다'
print()
print('⚠️ 판정: 단정 불가. 같은 문항으로 짝지은 비교와 p값을 요구할 것 (발표에 없음)')
```

## 검산대 ③ "수학 창발 +33점" (5·8장): 지표와 오염

의심 신호: 불균형 급등(+0.8 vs +33) · 전부-정답형 지표 · 3년 전 공개.
변형판(숫자·표현 교체) 점수를 요청해 받았다고 하자: **64점**.

```python
orig, variant, prev_gen = 91, 64, 58
drop = orig - variant
print(f'원본 {orig} vs 변형판 {variant} → 낙차 {drop}점, 강한 오염 신호(8장)')
print(f'정직한 실력 추정: 변형판 기준 {variant}점대 → 전 세대({prev_gen}) 대비 +{variant - prev_gen}점의 착실한 개선')

assert drop == 27 and variant - prev_gen == 6
print()
print('❌ 판정: "창발 +33점"은 오염 강한 의심. 실질 개선은 +6점 수준으로 추정')
```

## 검산대 ④ "대전 승률 55%" (6·9장): 표본과 심판

```python
n_battle, winrate = 200, 0.55
hw_b = half_width(winrate, n_battle)
z = (winrate - 0.5) / math.sqrt(0.25 / n_battle)
print(f'승률 55% ± {hw_b:.1f}%p → [{55 - hw_b:.1f}, {55 + hw_b:.1f}], 50%(반반) 포함')
print(f'z = {z:.2f} (< 1.96), 유의하지 않음')
print('+ 심판이 자사 형제 모델(자기 선호 미감정), 위치 무작위화 언급 없음')

assert round(hw_b, 1) == 6.9 and z < 1.96
print()
print('⚠️ 판정: 이중 보류 (판 수 부족 + 심판 미감정)')
```

신선한 홀드아웃, 같은 문항으로 짝지은 비교(p값), 독립 심판 + 사람 일치율, 전체 결과 공개.

핵심 통찰 둘. ① 이 발표엔 **수치 조작이 없었다**: 문제는 해석("숫자가 결론을 지지하나").

언어도 "성립하지 않음 · 요구"이지 "나쁜 모델"이 아니다.
